```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.datasets import fetch_california_housing, load_wine
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, plot_tree
```

CLASSIFICAZIONE (WINE)


```python
wine = load_wine()
X_wine, Y_wine = wine.data, wine.target
print(X_wine[:5])
print(Y_wine[:5])
```

```python
dt_classifier = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

dt_classifier.fit(X_wine, Y_wine)
y_pred_class = dt_classifier.predict(X_wine)
```

```python
print(f"DT Depth:\t{dt_classifier.get_depth()}")
print(f"DT #Leaves:\t{dt_classifier.get_n_leaves()}")
```

```python
plt.figure(figsize=(14,10))
plot_tree(
    dt_classifier,
    feature_names=wine.feature_names,
    class_names=wine.target_names,
    filled=True,
    fontsize=10
)
plt.title("Decision Tree Classifier")
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.scatter(range(len(Y_wine)), Y_wine, label = "Real", alpha=0.6)
plt.scatter(range(len(Y_wine)), y_pred_class, label="Prediction", alpha = 0.4)
plt.title("Decision Tree Classifier")
plt.xlabel("Samples")
plt.ylabel("Classes")
plt.legend()
plt.show()
```

```python
classificazione_accuracy = (y_pred_class == Y_wine).sum() / len(Y_wine)
print(f"Accuracy Classification:\t{round(classificazione_accuracy * 100, 2)}%\t({classificazione_accuracy})")
```
