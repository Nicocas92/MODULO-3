```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
```

```python
data = load_breast_cancer(as_frame=True)
df = data.frame

display(df.tail())
X = df[data.feature_names]
y = df["target"]
print(X.shape)

```

```python
model = LogisticRegression(max_iter = 5000)
model.fit(X,y)
```

```python
print(f"Intercetta: {model.intercept_[0]}")
print(f"Coefficienti principali (5):")
for name, coef in zip(data.feature_names[:5], model.coef_[0][:5]):
    print(f"{name}: {coef:.3f}")
    
```

```python
y_pred = model.predict(X)
```

```python
corrette = np.sum(y==y_pred)
tot = len(y)
accuracy = corrette/ tot
print(f"Etichette corrette: {corrette}/{tot}")
print(f"Accuracy (%): {accuracy * 100:.2f}%")

```

```python
plt.figure(figsize=(8,6))
plt.scatter(range(tot), y, label="Reali", alpha = 0.6)
plt.scatter(range(tot), y_pred, label = "Predetti", alpha = 0.6)
plt.title("Grafico Reali-Predette")
plt.xlabel("Indice campione")
plt.ylabel("Classe (0: benigno, 1: maligno)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize = (8,6))
plt.scatter(df["mean radius"],y, alpha = 0.4, label = "Reali")
plt.scatter(df["mean radius"], y_pred, alpha = 0.4, color = "red", label = "predette")
plt.xlabel("Mean radius")
plt.ylabel("Class")
plt.title("Relazione MeanRadius - Classe")
plt.grid()
plt.show()
```
